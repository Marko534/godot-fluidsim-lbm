## Godot Monitor CSV Columns — LBM Profiling Reference

### Default
#### Performance (most important for paper)

| Column | What it measures |
|---|---|
| `time/fps` | Frames per second |
| `time/process` | Time spent in `_process()` in seconds |
| `time/physics_process` | Time spent in `_physics_process()` |

#### GPU / Memory (critical for VRAM comparison)

| Column | What it measures |
|---|---|
| `video/video_mem` | Total VRAM usage |
| `video/texture_mem` | VRAM used by textures — 3D wind texture shows here |
| `video/buffer_mem` | VRAM used by buffers — F[], FPROP[], B[] show here |

#### Draw Calls

| Column | What it measures |
|---|---|
| `raster/total_draw_calls` | Total GPU draw calls per frame |
| `raster/total_primitives_drawn` | Total triangles drawn |
| `pipeline/compilations_*` | Shader compilations — should be 0 after warmup |

#### Key columns for D3Q15 vs D3Q19 vs D3Q27 comparison
- `time/fps`
- `time/process`
- `video/video_mem`
- `video/buffer_mem`

### Profiler

#### Frame Timing

| Column | What it measures |
|---|---|
| `category_frame_time` | Total frame time in seconds |
| `physics_frame_time` | Total physics frame time |
| `process_time` | Total time in all `_process()` calls |
| `physics_time` | Total time in all `_physics_process()` calls |

#### Engine Categories

| Column | What it measures |
|---|---|
| `categ::physics_3d` | Total 3D physics time |
| `categ::physics_3d::integrate_forces` | Force integration step |
| `categ::physics_3d::generate_islands` | Island generation step |
| `categ::physics_3d::setup_constraints` | Constraint setup |
| `categ::physics_3d::solve_constraints` | Constraint solver |
| `categ::physics_3d::integrate_velocities` | Velocity integration |
| `categ::physics_3d::flush_queries` | Physics query flushing |
| `categ::audio_thread::audio_server` | Audio server processing |
| `categ::audio_thread::audio_driver` | Audio driver overhead |

#### LBM Simulation (key columns for your paper)

| Column | What it measures |
|---|---|
| `res://main.gd::144::_process` | Total `_process()` per frame — main cost |
| `res://main.gd::149::_compute_process` | GPU compute dispatch time |
| `res://main.gd::165::_update_params` | Param buffer update cost |
| `res://main.gd::136::_run_init` | One-time LBM init cost |
| `res://main.gd::51::_create_buffers` | Buffer allocation cost |
| `res://main.gd::174::_create_boundary_from_noise` | Noise boundary generation cost |
| `res://main.gd::94::_make_uniform_set` | Uniform set creation cost |
| `res://main.gd::118::_setup_pipelines` | Pipeline setup cost |
| `res://main.gd::131::_load_shader` | Shader load/compile cost |

#### Player (can ignore for LBM benchmarking)

| `res://player/player.gd::233::_handle_headbob` | Headbob effect |
